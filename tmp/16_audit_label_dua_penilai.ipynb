# 16 · Audit label sebelum perbandingan XGBoost dan Temporal GNN
Tujuan audit adalah meninjau kasus prioritas, mengukur kesepakatan manusia pada sampel, dan membekukan versi anotasi untuk notebook 15. Penilai utama dan kedua menilai sendiri sebelum mendiskusikan perbedaan. Penilai kedua cukup menerima formulir HTML miliknya. File itu berisi koordinat pemain dan bola untuk visualisasi; gunakan sesuai hak akses data penelitian. Formulir dapat dibuka secara lokal tanpa koneksi internet.

```python
import sys
from pathlib import Path
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'src/protocol_label_review.py').is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from src.protocol_label_review import prepare, finalize

SOURCE = ROOT / 'artifacts/final_training_20261027T152829Z/frozen_annotations.csv'
PROCESSED = ROOT / 'data/processed'
OUT = ROOT / 'artifacts/audit_label_poin2'
```

## 1. Siapkan formulir lokal
Prioritas mencakup semua label ragu, keyakinan rendah, dan anotasi terakhir dari AI. Penilai utama mengisi kasus prioritas. Penilai kedua mengisi prioritas ditambah sampel acak lima episode per kelas per pertandingan dari kasus lainnya (seed 42). Untuk sampel acak, pembandingnya ialah label asli penilai manusia. Identitas pertemuan diketahui dari ID, tetapi label lama, model, fitur tekanan dan hasil perebutan bola disembunyikan. Bukti berupa animasi tracking dengan koordinat dibulatkan untuk visualisasi.

```python
jumlah, prioritas, sampel = prepare(SOURCE, PROCESSED, OUT)
print(f'{jumlah} episode. Penilai utama: {prioritas}. Penilai kedua: {prioritas + sampel}.')
print(OUT / 'penilai_utama.html')
print(OUT / 'penilai_kedua.html')
```

## 2. Kerjakan review secara terpisah

## 3. Selesaikan perbedaan dan bekukan label
Jalankan sel berikut setelah kedua JSON lengkap. Bila ada perbedaan, diskusikan episode di perbedaan.csv, isi kesepakatan.html, unduh kesepakatan.json ke folder yang sama, lalu ulangi sel. Simpan berkas jawaban sumber beserta log akhir.

```python
if (OUT / 'final/annotations.csv').exists():
    print('Versi final tersedia:', OUT / 'final/annotations.csv')
elif all((OUT / f'{role}.json').exists() for role in ('penilai_utama', 'penilai_kedua')):
    print(finalize(SOURCE, PROCESSED, OUT))
else:
    print('Menunggu dua lembar jawaban lengkap. Dataset final belum dibekukan.')
```
